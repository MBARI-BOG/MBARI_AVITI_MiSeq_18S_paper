## Run qiime2 fastq summary on

All reads combined, removing controls:
* (PLOT) Mean base quality at each position, with all reads combined
* % bases with median Qscore > 30, 35, 40


```python
%%bash
# For MiSeq data
python create_qiime2_manifest.py --directory raw_data/FD_20250210_BAK16171_18S_PE150 --output manifest_miseq_noControls.tsv

# For AVITI data  
python create_qiime2_manifest.py --directory raw_data/FK_20250312_BAK16171_18S_AVITI_PE150/20250312_BAK16171_18S_AVITI_PE150/L1_7pm --output manifest_aviti_noControls.tsv
```

```python
%%bash
qiime tools import \
    --type 'SampleData[PairedEndSequencesWithQuality]' \
    --input-path manifest_miseq_noControls.tsv \
    --output-path miseq_paired_end_sequences_with_quality_noControls.qza \
    --input-format PairedEndFastqManifestPhred33V2

qiime tools import \
    --type 'SampleData[PairedEndSequencesWithQuality]' \
    --input-path manifest_aviti_noControls.tsv \
    --output-path aviti_paired_end_sequences_with_quality_noControls.qza \
    --input-format PairedEndFastqManifestPhred33V2
    
    
```

```python
%%bash
qiime demux summarize \
    --i-data miseq_paired_end_sequences_with_quality_noControls.qza \
    --p-n 100000 \
    --o-visualization miseq_paired_end_sequences_with_quality_noControls.qzv

qiime demux summarize \
    --i-data aviti_paired_end_sequences_with_quality_noControls.qza \
    --p-n 100000 \
    --o-visualization aviti_paired_end_sequences_with_quality_noControls.qzv
```

```python
%%bash
unzip miseq_paired_end_sequences_with_quality_noControls.qzv -d miseq_noControls-fastq-summary
cp miseq_noControls-fastq-summary/*/data/forward-seven-number-summaries.tsv ./miseq_noControls-fastq-forward-seven.tsv
cp miseq_noControls-fastq-summary/*/data/reverse-seven-number-summaries.tsv ./miseq_noControls-fastq-reverse-seven.tsv
/bin/rm -rf miseq_noControls-fastq-summary

%%bash
unzip aviti_paired_end_sequences_with_quality_noControls.qzv -d aviti_noControls-fastq-summary
cp aviti_noControls-fastq-summary/*/data/forward-seven-number-summaries.tsv ./aviti_noControls-fastq-forward-seven.tsv
cp aviti_noControls-fastq-summary/*/data/reverse-seven-number-summaries.tsv ./aviti_noControls-fastq-reverse-seven.tsv
/bin/rm -rf aviti_noControls-fastq-summary
```
